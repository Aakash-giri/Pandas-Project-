```python
# Cleaning the dataframe

import pandas as pd 

df = pd.read_csv("Data.csv")

#Removing empty rows and columns 
df = df.dropna(how='all')
df = df.dropna(axis=1, how='all')

# Removing duplicates
df = df.drop_duplicates()

#Cleaning column names 

df.columns = df.columns.str.strip()

# Removing unnecessary index 
df = df.drop(columns=['Unnamed: 0'])


# reset index 
df = df.reset_index(drop='True')

# Display Cleaned data in a table

print(df.head(1377).to_string())

# Saving the cleaned data
df.to_csv("Cleaned_Data.csv",index=False)
```

```python
# 10. Check whether the dataset contains missing values.
# Find:

# Which columns contain missing values?
# How many missing values are present in each column?

import pandas as pd 

df = pd.read_csv("Cleaned_Data.csv")

missing_values = df.isnull().sum()

print("Missing Values are:")
print(missing_values[missing_values>0])
```

```python
#1. College count by ownership type
# Create a Matplotlib bar chart showing how many colleges belong to each Ownership Type.
# - X-axis: Ownership Type
# - Y-axis: Number of colleges
# - Add a title and axis labels.

import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns 

df = pd.read_csv("Cleaned_Data.csv")

Ownership_count = df['Ownership Type'].value_counts()
x = Ownership_count.index
y = Ownership_count.values
plt.figure(figsize=(9,5))  
plt.bar(x, y)
plt.title('Colleges counted by Onwership Type')
plt.xlabel('Ownership Type')
plt.ylabel('Number of Colleges')
plt.show()
```

```python
#2. Top 10 locations by college count
# Use a horizontal bar chart to display the 10 locations containing the most colleges.
# - Show location names on the Y-axis.
# - Show college counts on the X-axis.
# - Sort the bars from highest to lowest.

import pandas as pd 
import matplotlib.pyplot as plt 

df = pd.read_csv('Cleaned_Data.csv')
college_count = df['Location'].value_counts().head(10)
x = college_count.index
y = college_count.values
plt.barh(x,y)
plt.title('Top 10 locations by college count')
plt.xlabel('college counts')
plt.ylabel('locations')
plt.gca().invert_yaxis() #By default, plt.barh() places the first location at the bottom. 
plt.tight_layout()       # plt.gca().invert_yaxis() puts the location with the highest college count at the top, so the chart reads from highest to lowest
plt.show()
```

```python
#3. Percentage of colleges by ownership type
# Create a pie chart showing each ownership type's share of all college records.
# - Display percentage values.
# - Include a legend or category labels.
# - Give the chart a meaningful title.

import pandas as pd 
import matplotlib.pyplot as plt 
df = pd.read_csv("Cleaned_Data.csv")

Ownership_count = df["Ownership Type"].value_counts()
plt.figure(figsize=(10,7))
plt.pie(Ownership_count, labels=Ownership_count.index, autopct='%1.6f%%')
plt.title("Pie Chart Showing colleges by Ownership Type")
plt.legend(loc='upper left',bbox_to_anchor=(1,1))
plt.show()

#To adjust it even closer to the corner, try these values:
# Code	Position
# bbox_to_anchor=(0, 1)	Top-left corner
# bbox_to_anchor=(-0.05, 1.05)	Slightly outside, above-left
# bbox_to_anchor=(0.02, 0.98)	Slightly inside the corner
# bbox_to_anchor=(0, 1.05)	Above the top-left corner
```

```python
# 4. Count colleges using Seaborn
# Recreate Question 1 using Seaborn's countplot() instead of Matplotlib's plt.bar().
# - Use Ownership Type as the categorical variable.
# - Rotate labels if they overlap.
# - Add a title.
# Hint: sns.countplot()

import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns 

df = pd.read_csv('Cleaned_Data.csv')
sns.set_style("whitegrid")
plt.figure(figsize=(10,5))
sns.countplot(data=df, x='Ownership Type')
plt.xticks(rotation=30)
plt.title('Number of Colleges by Ownership Type')
plt.xlabel('Ownership Type')
plt.ylabel('Colleges')
plt.show()

```
