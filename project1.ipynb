```python
# Cleaning the dataframe

import pandas as pd 

df = pd.read_csv("Data.csv")

#Removing empty rows and columns 
df = df.dropna(how='all')
df = df.dropna(axis=1, how='all')

# Removing duplicates
df = df.drop_duplicates()

#Cleaning column names 

df.columns = df.columns.str.strip()

# Removing unnecessary index 
df = df.drop(columns=['Unnamed: 0'])


# reset index 
df = df.reset_index(drop='True')

# Display Cleaned data in a table

print(df.head(1377).to_string())

# Saving the cleaned data
df.to_csv("Cleaned_Data.csv",index=False)
```

```python
# 6. Find all colleges that offer BIT (Bachelor of Information Technology).
# Display their College and University.

import pandas as pd 

df = pd.read_csv("Cleaned_Data.csv")
print(df.columns)
print(df['Course Offered'])
BIT_college = df[df['Course Offered'].str.contains("BIT")]
print(BIT_college[['College','University']])
```

```python
# 7. Find all colleges that are affiliated with Tribhuvan University.
# Display:

# College name
# Location
# Ownership Type

import pandas as pd

df =pd.read_csv("Cleaned_Data.csv")
print(df.columns)
TU = df[df["University"].str.contains("Tribhuvan University")]
print(TU[["College","Location","Ownership Type"]])

```

```python
# 8. Find the number of colleges in each location.
# Display the 10 locations with the highest number of colleges.

import pandas as pd 

df = pd.read_csv("Cleaned_Data.csv")

location = df.groupby("Location").size()
print(location.nlargest(10))
```

```python
# 9. Find colleges whose names contain the word "College".
# Make the search case-insensitive.

import pandas as pd 

df = pd.read_csv("Cleaned_Data.csv")
College_names = df[df['College'].str.contains('College',case=False)]
print(College_names)
```

```python
# 10. Check whether the dataset contains missing values.
# Find:

# Which columns contain missing values?
# How many missing values are present in each column?

import pandas as pd 

df = pd.read_csv("Cleaned_Data.csv")

missing_values = df.isnull().sum()

print("Missing Values are:")
print(missing_values[missing_values>0])
```

```python
# # Advanced Questions 

# 11. Find colleges that offer both BIT and BCA.

# Your result should contain only:

# College
# Location
# University

# 💡 Hint: You'll need to apply multiple conditions to the same Course Offered column.

import pandas as pd 

df = pd.read_csv("Cleaned_Data.csv")
Courses = df[
    df['Course Offered'].str.contains('BIT',case=False,na=False) &
    df['Course Offered'].str.contains('BCA',case=False,na=False)
]
print(Courses[['College','Location','University']])



```
