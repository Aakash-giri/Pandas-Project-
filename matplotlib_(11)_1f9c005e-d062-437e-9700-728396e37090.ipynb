The function plt.hist() in Matplotlib is used to create a histogram, which shows the distribution of numerical data.

What is a Histogram?
A histogram divides the range of values into intervals called bins.

It then counts how many values fall into each bin and represents them as bars.

Unlike a bar chart (which is for categorical data), a histogram is for continuous or numerical data.

```python
import matplotlib as plt
data = [1,2,3,4,5,1,2,3,6,7,8,1,2,3,5.5, 6.5]
plt.hist(data, edgecolor = "red", color="black", bins=10)
plt.show()
```

Pie Chart

What it is:
A circle divided into slices, where each slice shows a percentage or fraction of the whole.

When to use:
To show proportions (e.g., "Market share of smartphone companies").

```python
labels = ['Apple', 'Samsung', 'Xiaomi', 'Redmi','Others']
sizes = [30, 25, 20, 25, 43]
# explode = [0, 0, 0.1, 0.1]
plt.figure(figsize=(10,5))
# plt.pie(sizes, labels=labels, autopct='%1.5f%%', explode=[0.2,0,0.1,0,0])
plt.pie(sizes, labels=labels, autopct='%1.5f%%', explode=[0.2,0,0.1,0,0])
# Explode brings out a part out from the piechart
plt.title("Pie Chart")  # To make it a circle
plt.show()
```


3d Flow:
1. Create a blank figure => fig = plt.figure()
2. Add subplot in blank figure => fig.add_subplot()
3. keep the prijection 3d => abc = fig.add_subplot(projection="3d)

```python
x = np.random.rand(50)
y = np.random.rand(50)
z = np.random.rand(50)

# 3D FLow
# 1. Create a blank figure
# 2. Fig.add_subplot(projection='3d')

fig = plt.figure() # Creates a new blank figure (container for plots).
ax = fig.add_subplot(projection="3d") # Adds a 3D plotting area (subplot) to the figure.The argument projection='3d' is what makes this a 3D plot.
ax.scatter(x,y,z, label='Ujjwal Neupane') # Creates a 3D scatter plot with the points (x, y, z).
ax.set_title('3D Line Plot Example')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
plt.legend(fontsize=10, loc="upper left", title='3d scatter plot')
plt.show()

# Plotly
# Task -  Add a title for this plot, add x-axis, y-axis, and z-axis label also
```

Stack Plot

What it is:
A line plot where multiple datasets are stacked on top of each other.
It shows how different categories contribute to the total over time.

days → x-axis values.

sleeping, eating, working, playing → multiple datasets stacked on top of each other.

labels → Names for each dataset (used in the legend).

colors → Colors for each stack:

'm' = magenta (Sleep)

'c' = cyan (Eat)

'r' = red (Work)

'k' = black (Play)

Syntax: plt.stackplot(x, y1, y2, y3, ...)

x → values for the horizontal axis

y1, y2, ... → each array represents a layer of the stack (vertical values stacked on top of each other)

Optional: colors=..., labels=...


```python
days = ["Sunday","Monday","Tuesday","Wednesday","Thursday", "Friday"]
sleeping = [7.5, 6, 6, 11, 7, 5]
eating = [2, 3, 4, 3, 2, 3]
working = [2, 8, 7, 2, 2, 6]
playing = [8, 5, 7, 8, 13,11]
writing = [8, 5, 7, 8, 13, 17]
gaming = [-2, 5, 7, 8, 13, 13]
coding = [8, 5, 7, 8, 13, 4]

plt.stackplot(days,sleeping, eating, working, playing,writing,gaming,coding,
              labels=['Sleep', 'Eat', 'Work', 'Play', 'Write', 'Game', 'Code'],
              colors=['c','b','r','black','purple','orange','green'])
plt.legend(loc='lower right')
plt.title('Stack Plot Example')
plt.xlabel('Days')
plt.ylabel('Hours')
plt.xticks(rotation=90) # Set x-ticks to match the days
plt.show()

```

Box Plot (Whisker Plot)

What it is:
Shows the spread of data:

The box shows the middle 50% of data (quartiles).

The line inside the box is the median.

Whiskers show the min and max values.

Dots outside whiskers are outliers.

What Does a Box Plot Show?

Median (Q2): The middle value of the dataset.

Q1 (25th percentile): The value below which 25% of the data lies.

Q3 (75th percentile): The value below which 75% of the data lies.

Interquartile Range (IQR): Q3 - Q1, the "height" of the box.

Whiskers: Extend to the smallest and largest values within 1.5 × IQR.

Outliers: Data points beyond whiskers, shown as dots or circles.

When to use:

A heatmap is a visualization that uses colors to show relationships, patterns, and correlations in data, making complex numbers easy to understand at a glance.

```python
import matplotlib.pyplot as plt

# X-axis categories
quarters = ['Q1', 'Q2', 'Q3', 'Q4']

# Heights of bars for Product A
product_a = [20, 35, 30, 25]

# Heights of bars for Product B
product_b = [15, 25, 35, 30]

# Plot the first set of bars
plt.bar(
    quarters,          # X-axis categories
    product_a,         # Heights of bars
    label='Product A', # Label for legend
    color='skyblue',   # Bar color (optional)
    width=0.5,         # Width of bars (optional)
    edgecolor='black'  # Edge color of bars (optional)
)

# Plot the second set of bars stacked on top of Product A
plt.bar(
    quarters,
    product_b,
    bottom=product_a,   # Stack on top of product_a
    label='Product B',
    color='orange',
    width=0.5,
    edgecolor='black',
    alpha = 0.5
)

# Add title of the chart
plt.title("Quarterly Sales", fontsize=14)

# Label for Y-axis
plt.ylabel("Sales", fontsize=12)

# Display legend
plt.legend()

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Create 100 points evenly spaced between 0 and 2*pi (full circle)
theta = np.linspace(0, 2*np.pi, 100)

# Define the radius for each angle (r = sin(theta))
r = np.sin(theta)

# Create a polar plot
plt.figure(figsize=(5,5))
plt.polar(
    theta,  # Angles (in radians)
    r,      # Radius values corresponding to each angle
    color='r',      # Color of the line (optional)
    linestyle='-',     # Line style: '-', '--', '-.', ':' (optional)
    linewidth=2,       # Width of the line (optional)      # Marker style for points (optional)
        # Size of the markers (optional)
    label='r = sin(theta)',# Label for legend (optional)
)

# Add a title to the plot
plt.title("Polar Plot", fontsize=14)

# Add legend if labels are provided
plt.legend()

# Show the plot
plt.show()
```
